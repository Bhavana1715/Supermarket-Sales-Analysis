```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df1=pd.read_csv('/content/drive/MyDrive/supermarket_sales/annex1.csv')
df2=pd.read_csv('/content/drive/MyDrive/supermarket_sales/annex2.csv')
df3=pd.read_csv('/content/drive/MyDrive/supermarket_sales/annex3.csv')
df4=pd.read_csv('/content/drive/MyDrive/supermarket_sales/annex4.csv')
display(df1.head(10))
display(df2.head(10))
display(df3.head(10))
display(df4.head(10))
```

```python
merged_df = pd.merge(df1, df2, on='Item Code', how='left')
merged_df['Total Revenue']=merged_df['Quantity Sold (kilo)'] * merged_df['Unit Selling Price (RMB/kg)']
merged_df.head(10)
```

**Top 10 Items by Total Revenue**

```python
merged_df['Total Revenue'] = merged_df['Quantity Sold (kilo)'] * merged_df['Unit Selling Price (RMB/kg)']

top_items = merged_df.groupby('Item Name')['Total Revenue'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.bar(top_items.index, top_items.values, color='green')
plt.xticks(rotation=45)
plt.title('Top 10 Items by Total Revenue')
plt.ylabel('Revenue (RMB)')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Get the underperforming items
below_items = (merged_df.groupby('Item Name')['Total Revenue'].sum().sort_values(ascending=True).head(10))


# Use numeric x-values instead of string names directly
x = range(len(below_items))
labels = below_items.index
values = below_items.values

plt.figure(figsize=(10, 6))
plt.bar(x, values, color='red')

plt.xticks(x, labels, rotation=45, ha='right')  # set tick labels manually
plt.title('Under Performers')
plt.ylabel('Revenue (RMB)')
plt.tight_layout()
plt.xlim(-0.5, len(below_items) - 0.5)  # flush bars to left and right
plt.show()
```

```python
# Assuming you already loaded the wholesale price table:
# Let's name it 'wholesale_df'
# Columns: 'Date', 'Item Code', 'Wholesale Price (RMB/kg)'

# Ensure date formats are consistent
merged_df['Date'] = pd.to_datetime(merged_df['Date'])
df3['Date'] = pd.to_datetime(df3['Date'])

# Merge on both Item Code and Date
merged_df = merged_df.merge(df3,on=['Item Code', 'Date'],how='left')  # keep all sales even if wholesale price is missing

# Check if merge worked
print(merged_df[['Item Code', 'Date', 'Unit Selling Price (RMB/kg)', 'Wholesale Price (RMB/kg)']].head())
```

```python
# Step 1: Compute Profit per kg
merged_df['Profit per kg'] = merged_df['Unit Selling Price (RMB/kg)'] - merged_df['Wholesale Price (RMB/kg)']

# Step 2: Compute Total Profit
merged_df['Total Profit'] = merged_df['Profit per kg'] * merged_df['Quantity Sold (kilo)']

# Step 3: Group by Item and calculate total revenue & total profit
profit_vs_sales = merged_df.groupby('Item Name').agg({
    'Total Revenue': 'sum',
    'Total Profit': 'sum'
}).sort_values(by='Total Revenue', ascending=False)

# Step 4: Plot
plt.figure(figsize=(10, 6))
plt.scatter(profit_vs_sales['Total Revenue'], profit_vs_sales['Total Profit'], color='purple')
plt.xlabel('Total Revenue')
plt.ylabel('Total Profit')
plt.title('Revenue vs Profit: Are High Sellers Also High Earners?')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### **✅ Key Observations:**
**Positive Correlation:**
There is a clear upward trend — as revenue increases, profit generally increases.

➤ This suggests that most high-selling items also tend to be high-profit items.

**Clustered Bottom Left:**
A large number of items are concentrated in the low revenue–low profit zone.

➤ These are likely underperformers or niche products.

```python
# Step 1: Merge df4 into merged_df on the common column (likely 'Item Name')
# Make sure the column names match exactly

final_df = merged_df.merge(df4[['Item Name', 'Loss Rate (%)']], on='Item Name', how='left')
```

```python
# Step 2: Group and prepare for plotting
loss_vs_profit = final_df.groupby('Item Name')[['Loss Rate (%)', 'Total Profit']].mean().reset_index()

# Step 3: Visualize
plt.figure(figsize=(10, 6))
plt.scatter(loss_vs_profit['Loss Rate (%)'], loss_vs_profit['Total Profit'], color='crimson', alpha=0.7)
plt.title('Impact of Loss Rate on Item Profitability')
plt.xlabel('Loss Rate (%)')
plt.ylabel('Average Profit (RMB)')
plt.grid(True)
plt.tight_layout()
plt.show()
display(final_df.head(5))
```

## **NO Strong Correlation:**

The dots are fairly scattered horizontally, suggesting that there's no clear linear relationship between Loss Rate (%) and Average Profit.

Even items with 0% to 10% loss rate show a wide range of profitability, from negative to slightly positive

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming final_df is your merged dataframe

# 1. Group by Discount and evaluate profitability
discount_analysis = final_df.groupby('Discount (Yes/No)') \
    .agg({'Profit per kg': 'mean', 'Total Profit': 'sum', 'Total Revenue': 'sum'}) \
    .reset_index()

print("Discount Analysis:\n", discount_analysis)

# 2. Compare Items With and Without Discounts
with_discount = final_df[final_df['Discount (Yes/No)'] == 'Yes']
without_discount = final_df[final_df['Discount (Yes/No)'] == 'No']

with_discount_summary = with_discount[['Item Name', 'Total Revenue', 'Total Profit']].groupby('Item Name').sum()
without_discount_summary = without_discount[['Item Name', 'Total Revenue', 'Total Profit']].groupby('Item Name').sum()

print("\nWith Discount:\n", with_discount_summary)
print("\nWithout Discount:\n", without_discount_summary)

# 3. Visualize Discount Impact on Profitability
profit_comparison = final_df.groupby('Discount (Yes/No)')['Profit per kg'].mean()

# Bar chart
plt.figure(figsize=(8, 6))
profit_comparison.plot(kind='bar', color=['green', 'red'])
plt.title('Average Profit per Item: With vs Without Discount')
plt.ylabel('Average Profit per kg')
plt.xticks(rotation=0)
plt.show()

# 4. Correlation Between Discounts and Profitability
final_df['Discount (Binary)'] = final_df['Discount (Yes/No)'].map({'Yes': 1, 'No': 0})

# Calculate correlation
correlation_profit = final_df[['Discount (Binary)', 'Profit per kg']].corr()
print("\nCorrelation between Discount and Profitability:\n", correlation_profit)
```



### **1. Higher Profit Without Discount**

   * Items sold *without a discount* have a significantly higher *average profit per kg* (over 3 units).
   * Items sold *with a discount* yield a much *lower average profit*, approximately half (around 1.5 units or less).

### **2. Discount Impact:**

   * Applying discounts clearly reduces the average profit per item.
   * This implies that while discounts might drive sales volume, they significantly cut into profit margins.
